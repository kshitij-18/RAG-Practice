```python
from pathlib import Path
from llama_index.core import SimpleDirectoryReader
```

```python
notes_dir = "../coffee_pages"
path = Path(notes_dir)
# Converting Rawdata into Docs
docs = SimpleDirectoryReader(str(path)).load_data()
```

```python
len(docs)
```

```python
docs[0]
```

```python
# From documents to Nodes
from llama_index.core.node_parser import SimpleNodeParser

# 2) NODE / CHUNK CONSTRUCTION
parser = SimpleNodeParser.from_defaults(
    chunk_size=700,   # tweak here
    chunk_overlap=70  # tweak here
)

nodes = parser.get_nodes_from_documents(docs)
print("nodes built:", len(nodes))
```

```python
nodes[0].__dict__
```

```python
nodes[1].__dict__
```

# Index

```python
from llama_index.core import ListIndex

index_list = ListIndex(nodes)
```

```python
print("index built:", type(index_list))

query = "fusion coffee"

retriever = index_list.as_retriever()
results = retriever.retrieve(query)
```

```python
#results
```

```python
print("retrieved:", len(results))
for i, n in enumerate(results, 1):
    print(f"--- RESULT {i} ---")
    print("metadata:", n.metadata)
    print("text:", n.get_content()[:200])
```

# KeyWord Index

```python
import os
from llama_index.core import (
    SimpleDirectoryReader,
    Settings,
    SummaryIndex,
    TreeIndex,
    KeywordTableIndex
)
```

```python
import os
from dotenv import load_dotenv

load_dotenv()

gemini_api_key = os.getenv("GEMINI_API_KEY")
pinecone_api_key = os.getenv("PINECONE_API_KEY")
```

```python
# Updated imports
from llama_index.llms.gemini import Gemini
from llama_index.core import Settings


# Configure the LLM
Settings.llm = Gemini(
    model="gemini-2.5-flash", # Note: use the formal model string
    api_key=gemini_api_key,
    temperature=0,
)
```

```python
kw_index = KeywordTableIndex.from_documents(docs)
```

```python
type(kw_index)
```

```python
# table could be .table or .keyword_table depending on version
idx_struct = kw_index.index_struct
print("=== KEYWORD INDEX STRUCT TYPE ===")
print(type(idx_struct))

# table could be .table or .keyword_table depending on version
table = getattr(idx_struct, "table", None)
if table is None:
    table = getattr(idx_struct, "keyword_table", None)

if table is None:
    print("No keyword table attribute found on this version.")
else:
    print("\n=== KEYWORD → ENTRIES (first 20) ===")
    for i, (kw, entries) in enumerate(table.items()):
        print(f"{kw!r} -> {len(entries)} entries")
        if i >= 19:
            break
```

```python
query = "spices coffee"

# Step 1: let retriever find matching nodes
retriever = kw_index.as_retriever(similarity_top_k=5)
results = retriever.retrieve(query)

# Step 2: collect matched node IDs
matched_node_ids = {r.node.node_id for r in results}
matched_node_ids
```

```python


# Get keyword table (keyword -> [node_ids])
keyword_table = kw_index.index_struct.table

for i, r in enumerate(results, 1):
    node = r.node
    node_id = node.node_id

    print(f"\n===== RESULT {i} =====")

    # ---- Print text ----
    print("TEXT:")
    print(node.get_content()[:400])  # trim for readability

    # ---- Find matching keywords ----
    matched_keywords = [
        kw for kw, node_ids in keyword_table.items()
        if node_id in node_ids
    ]

    print("\nKEYWORDS:")
    if matched_keywords:
        print(matched_keywords)
    else:
        print("No keywords found")
```
